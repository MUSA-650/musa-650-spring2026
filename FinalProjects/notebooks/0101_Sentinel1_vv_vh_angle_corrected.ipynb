## Part 1: Data Acquisition + Preprocessing

We'll pull Sentinel-1 Level-1C Ground Range Detected (GRD) data over Nairobi from May 2024

Sentinel-1 is a pair of Synthetic Aperture Radar (SAR) imaging satellites launched in 2014 and 2016 by the European Space Agency (ESA). Their 6 day revisit cycle and ability to observe through clouds makes this dataset perfect for sea and land monitoring, emergency response due to environmental disasters, and economic applications. This dataset represents the global Sentinel-1 GRD archive, from beginning to the present, converted to cloud-optimized GeoTIFF format.

```python
# Define spatial and temporal extent
dx = 0.0003  # ~30m resolution
epsg = 4326

# Nairobi area bounds
#-1.5371°, 36.5134° : -1.0918°, 37.2744°
latmin, latmax = -1.5371, -1.0918
lonmin, lonmax = 36.5134, 37.2744
bounds = (lonmin, latmin, lonmax, latmax)

# Temporal extent: May 2026
#start_date = datetime(year=2026, month=3, day=1)
#end_date = start_date + timedelta(days=20)

# Temporal extent: April 2024
start_date = datetime(year=2024, month=3, day=6)
end_date = datetime(year=2024, month=6, day=1)

time_format = "%Y-%m-%d"
date_query = f"{start_date.strftime(time_format)}/{end_date.strftime(time_format)}"

# Search for Sentinel-1 data
items = (
    pystac_client.Client.open("https://earth-search.aws.element84.com/v1")
    .search(
        bbox=bounds,
        collections=["sentinel-1-grd"],
        datetime=date_query,
        limit=100,
        query={"sar:instrument_mode": {"in": ["IW"]}},
    )
    .item_collection()
)
print(f"{len(items)} scenes found")
```

### Angle / Stripe-Aware Raw Preprocessing

This section converts Sentinel-1 backscatter to dB and removes broad column-wise brightness bias before baseline differencing. This is an empirical replacement for incidence-angle correction because the Earth Search `sentinel-1-grd` STAC items used here do not provide an `angle` raster asset.


```python
# Preprocess raw Sentinel-1 before computing the flood baseline.
# This reduces the vertical brightness bands caused by incidence angle / swath geometry.

def _spatial_dims(da):
    non_time_dims = [d for d in da.dims if d != "time"]
    if len(non_time_dims) != 2:
        raise ValueError(f"Expected one time dim and two spatial dims, got {da.dims}")
    y_dim = "y" if "y" in da.dims else non_time_dims[0]
    x_dim = "x" if "x" in da.dims else non_time_dims[1]
    return y_dim, x_dim


def to_db(da):
    """Convert positive linear backscatter to dB."""
    return 10 * np.log10(da.where(da > 0))


def remove_column_brightness_bias(da_db, window=121):
    """
    Empirical incidence-angle / stripe normalization.

    For each acquisition, estimate the large-scale x-direction brightness profile
    using the median over rows. Subtract that profile while preserving the scene
    median. This reduces vertical bands before before/after differencing.
    """
    y_dim, x_dim = _spatial_dims(da_db)
    scene_median = da_db.median(dim=(y_dim, x_dim), skipna=True)
    column_profile = da_db.median(dim=y_dim, skipna=True)
    smooth_profile = column_profile.rolling({x_dim: window}, center=True, min_periods=1).median()
    correction = smooth_profile - scene_median
    return da_db - correction


dc_db = xr.Dataset(
    {
        "vv": to_db(dc_raw.vv),
        "vh": to_db(dc_raw.vh),
    },
    attrs=dc_raw.attrs,
)

dc = xr.Dataset(
    {
        "vv": remove_column_brightness_bias(dc_db.vv),
        "vh": remove_column_brightness_bias(dc_db.vh),
    },
    attrs=dc_raw.attrs,
)

before = dc.sel(time=slice("2024-03-06", "2024-04-06"))
after = dc.sel(time=slice("2024-05-01", "2024-06-01"))
before_raw = dc_raw.sel(time=slice("2024-03-06", "2024-04-06"))
after_raw = dc_raw.sel(time=slice("2024-05-01", "2024-06-01"))

print(f"Before scenes: {before.sizes.get('time', 0)}")
print(f"After scenes: {after.sizes.get('time', 0)}")
dc
```

### Create a Binary Flood Mask

We'll now create a binary mask to identify potential flood areas. Flooding typically causes a significant decrease in SAR backscatter (VH in this case) due to specular reflection from smooth water surfaces. We'll set a threshold on the `difference` map (After / Before VH) to isolate these areas. A negative threshold value means we are looking for areas where VH has decreased.

```python
# Corrected dB difference: flooding/open water is usually a negative backscatter change.
flood_threshold = -1.5
flood_mask = difference < flood_threshold

plt.figure(figsize=(10, 8))
flood_mask.plot(cmap="Blues", add_colorbar=False)
plt.title(f"Binary Flood Mask from Corrected VH Difference (< {flood_threshold} dB)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

### Determining Flood Threshold using Otsu's Method

Otsu's method is a popular approach for automatically performing histogram-based image thresholding. It aims to find the threshold that minimizes the intra-class variance (the variance within the foreground and background classes) or, equivalently, maximizes the inter-class variance (the variance between the two classes). In our case, it will help us automatically distinguish between 'flooded' and 'non-flooded' areas based on the `difference` in VH backscatter.

```python
difference_flat = difference.values.flatten()
difference_flat = difference_flat[~np.isnan(difference_flat)]

Q1 = np.percentile(difference_flat, 25)
Q3 = np.percentile(difference_flat, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

difference_filtered = difference_flat[(difference_flat >= lower_bound) & (difference_flat <= upper_bound)]
otsu_threshold = threshold_otsu(difference_filtered)

print(f"Otsu's threshold for corrected VH dB difference: {otsu_threshold:.4f}")
flood_mask_otsu = difference < otsu_threshold

plt.figure(figsize=(10, 8))
flood_mask_otsu.plot(cmap="Blues", add_colorbar=False)
plt.title(f"Binary Flood Mask (Corrected VH Otsu threshold: {otsu_threshold:.2f} dB)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```

### Determining Flood Threshold using Otsu's Method (VV Band)

We will now apply Otsu's method to the `difference_vv` map to automatically find a suitable threshold for flood detection in the VV band. This method helps to minimize intra-class variance and maximize inter-class variance, providing an objective way to distinguish between flooded and non-flooded areas.

```python
difference_vv_flat = difference_vv.values.flatten()
difference_vv_flat = difference_vv_flat[~np.isnan(difference_vv_flat)]

Q1_vv = np.percentile(difference_vv_flat, 25)
Q3_vv = np.percentile(difference_vv_flat, 75)
IQR_vv = Q3_vv - Q1_vv
lower_bound_vv = Q1_vv - 1.5 * IQR_vv
upper_bound_vv = Q3_vv + 1.5 * IQR_vv

difference_vv_filtered = difference_vv_flat[(difference_vv_flat >= lower_bound_vv) & (difference_vv_flat <= upper_bound_vv)]
otsu_threshold_vv = threshold_otsu(difference_vv_filtered)

print(f"Otsu's threshold for corrected VV dB difference: {otsu_threshold_vv:.4f}")
flood_mask_otsu_vv = difference_vv < otsu_threshold_vv

plt.figure(figsize=(10, 8))
flood_mask_otsu_vv.plot(cmap="Blues", add_colorbar=False)
plt.title(f"Binary Flood Mask (Corrected VV Otsu threshold: {otsu_threshold_vv:.2f} dB)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.tight_layout()
plt.show()
```
