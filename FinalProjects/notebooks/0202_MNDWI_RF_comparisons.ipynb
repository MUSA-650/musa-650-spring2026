## Part 1: Data Acquisition + Preprocessing

We'll pull Sentinel-1 Level-1C Ground Range Detected (GRD) data over Nairobi from May 2024

Sentinel-1 is a pair of Synthetic Aperture Radar (SAR) imaging satellites launched in 2014 and 2016 by the European Space Agency (ESA). Their 6 day revisit cycle and ability to observe through clouds makes this dataset perfect for sea and land monitoring, emergency response due to environmental disasters, and economic applications. This dataset represents the global Sentinel-1 GRD archive, from beginning to the present, converted to cloud-optimized GeoTIFF format.

```python
# Define spatial and temporal extent
dx = 0.0003  # ~30m resolution
epsg = 4326

# Nairobi area bounds
#-1.5371°, 36.5134° : -1.0918°, 37.2744°
latmin, latmax = -1.5371, -1.0918
lonmin, lonmax = 36.5134, 37.2744
bounds = (lonmin, latmin, lonmax, latmax)

# Temporal extent: May 2026
#start_date = datetime(year=2026, month=3, day=1)
#end_date = start_date + timedelta(days=20)

# Temporal extent: April 2024
start_date = datetime(year=2024, month=2, day=1)
end_date = start_date + timedelta(days=180)

time_format = "%Y-%m-%d"
date_query = f"{start_date.strftime(time_format)}/{end_date.strftime(time_format)}"

# Search for Sentinel-2 data
items = (
    pystac_client.Client.open("https://earth-search.aws.element84.com/v1")
    .search(
        bbox=bounds,
        collections=["sentinel-1-grd"],
        datetime=date_query,
        limit=100,
        query={"sar:instrument_mode": {"in": ["IW"]}},
    )
    .item_collection()
)
print(f"{len(items)} scenes found")
```

```python
import os
os.environ['AWS_NO_SIGN_REQUEST'] = 'YES'
os.environ['GDAL_DISABLE_READDIR_ON_OPEN'] = 'EMPTY_DIR'
os.environ['GDAL_HTTP_MERGE_CONSECUTIVE_RANGES'] = 'YES'
os.environ['GDAL_HTTP_TIMEOUT'] = '300'
os.environ['GDAL_HTTP_MAX_RETRY'] = '5'
os.environ['GDAL_HTTP_RETRY_DELAY'] = '2'
os.environ['CPL_VSIL_CURL_ALLOWED_EXTENSIONS'] = '.tif,.tiff'

geobox = GeoBox.from_bbox(bounds, crs=f"epsg:{epsg}", resolution=dx)

# Load both VV and VH together so angle correction can be applied to both at once
dc_raw = odc.stac.load(
    items,
    bands=["vv", "vh"],
    chunks={"time": 5, "x": 600, "y": 600},
    geobox=geobox,
    resampling="bilinear",
)


def _spatial_dims(da):
    non_time_dims = [d for d in da.dims if d != "time"]
    if len(non_time_dims) != 2:
        raise ValueError(f"Expected one time dim and two spatial dims, got {da.dims}")
    y_dim = "y" if "y" in da.dims else non_time_dims[0]
    x_dim = "x" if "x" in da.dims else non_time_dims[1]
    return y_dim, x_dim


def to_db(da):
    """Convert positive linear backscatter to dB."""
    return 10 * np.log10(da.where(da > 0))


def remove_column_brightness_bias(da_db, window=121):
    """
    Empirical incidence-angle / stripe normalization.

    Sentinel-1 GRD items from Earth Search do not expose an incidence-angle
    raster, so this removes the broad column-wise brightness gradient caused by
    swath geometry. For each acquisition, the large-scale x-direction profile
    (median over rows) is smoothed and subtracted while preserving the scene median.
    """
    y_dim, x_dim = _spatial_dims(da_db)
    scene_median = da_db.median(dim=(y_dim, x_dim), skipna=True)
    column_profile = da_db.median(dim=y_dim, skipna=True)
    smooth_profile = column_profile.rolling({x_dim: window}, center=True, min_periods=1).median()
    correction = smooth_profile - scene_median
    return da_db - correction


dc_db = xr.Dataset(
    {"vv": to_db(dc_raw.vv), "vh": to_db(dc_raw.vh)},
    attrs=dc_raw.attrs,
)

dc = xr.Dataset(
    {
        "vv": remove_column_brightness_bias(dc_db.vv),
        "vh": remove_column_brightness_bias(dc_db.vh),
    },
    attrs=dc_raw.attrs,
)
dc
```

Visualizing the Raw SAR time series

```python
from scipy.ndimage import gaussian_filter1d
import numpy as np
smoothed_before = xr.apply_ufunc(
    gaussian_filter1d,
    before,
    kwargs={'sigma': 2, 'axis': list(before.dims).index('time')},
    dask='parallelized'
)
smoothed_after = xr.apply_ufunc(
    gaussian_filter1d,
    after,
    kwargs={'sigma': 2, 'axis': list(after.dims).index('time')},
    dask='parallelized'
)
```

```python
before_mean = smoothed_before.mean(dim="time")
after_mean = smoothed_after.mean(dim="time")
```

Flood Mask based off the UN-SPIDER Method

```python
# dB difference: negative values indicate backscatter decrease (open water / flooding)
difference = after_mean - before_mean
```

```python
# Pixels with > 1.5 dB decrease in VH backscatter are flagged as flooded.
# Specular reflection from open water lowers SAR backscatter, producing negative dB differences.
threshold = -1.5  # dB
difference_binary = (difference < threshold).astype(int)
```

### MNDWI Before / After Flood

MNDWI = (Green − SWIR16) / (Green + SWIR16). Positive values (blue) indicate open water; negative values (red) indicate vegetation or bare land. A positive shift in the difference map (After − Before) flags areas that became wetter or flooded.

```python
# Sentinel-2 time windows matching the SAR before/after periods
before_s2 = dc_valid.sel(time=slice("2024-03-06", "2024-04-06"))
after_s2  = dc_valid.sel(time=slice("2024-05-01", "2024-06-01"))

# Median composites for each period
before_green  = before_s2["green"].median(dim="time").compute()
before_swir16 = before_s2["swir16"].median(dim="time").compute()
after_green   = after_s2["green"].median(dim="time").compute()
after_swir16  = after_s2["swir16"].median(dim="time").compute()

# MNDWI = (Green - SWIR16) / (Green + SWIR16)
mndwi_before = (before_green - before_swir16) / (before_green + before_swir16)
mndwi_after  = (after_green  - after_swir16)  / (after_green  + after_swir16)

# Threshold-based water masks: pixels above threshold are classified as water
mndwi_threshold = 0.0
water_before = mndwi_before > mndwi_threshold
water_after  = mndwi_after  > mndwi_threshold
flood        = water_after & (~water_before)

fig, axes = plt.subplots(1, 3, figsize=(24, 8))

mndwi_before.plot(ax=axes[0], cmap="RdBu", vmin=-1, vmax=1)
water_before.where(water_before).plot(
    ax=axes[0], cmap="Blues", vmin=0, vmax=1, add_colorbar=False, alpha=0.4
)
axes[0].set_title(f"MNDWI Before Flood (Mar–Apr 2024)\nBlue = water (MNDWI > {mndwi_threshold})")

mndwi_after.plot(ax=axes[1], cmap="RdBu", vmin=-1, vmax=1)
water_after.where(water_after).plot(
    ax=axes[1], cmap="Blues", vmin=0, vmax=1, add_colorbar=False, alpha=0.4
)
axes[1].set_title(f"MNDWI After Flood (May–Jun 2024)\nBlue = water (MNDWI > {mndwi_threshold})")

mndwi_after.plot(ax=axes[2], cmap="gray", add_colorbar=False)
flood.where(flood).plot(
    ax=axes[2], cmap="Oranges", vmin=0, vmax=1, add_colorbar=False, alpha=0.8
)
axes[2].set_title("Flood Extent (MNDWI-based)\nOrange = water after, dry before")

for ax in axes:
    ax.set_xlabel("Longitude")
    ax.set_ylabel("Latitude")
    ax.set_aspect("equal")

legend_elements = [
    Patch(facecolor="#4393c3", alpha=0.6, label=f"Water (MNDWI > {mndwi_threshold})"),
    Patch(facecolor="orange",  alpha=0.8, label="Flood (new water after flood)"),
]
axes[2].legend(handles=legend_elements, loc="lower left", fontsize=8)

plt.suptitle("Sentinel-2 MNDWI — Before / After / Flood Extent", fontsize=14, y=1.01)
plt.tight_layout()
plt.show()

print(f"Water pixels before flood: {int(water_before.sum()):,}")
print(f"Water pixels after flood:  {int(water_after.sum()):,}")
print(f"Newly flooded pixels:      {int(flood.sum()):,}")
```

```python
import base64
import io

import folium
import matplotlib.colors as mcolors
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image


def da_to_image_overlay(da, cmap, name, vmin=None, vmax=None, opacity=0.6):
    """Render a 2-D xarray DataArray as a folium ImageOverlay."""
    data = da.values.astype(float)
    vmin = vmin if vmin is not None else np.nanmin(data)
    vmax = vmax if vmax is not None else np.nanmax(data)

    norm = mcolors.Normalize(vmin=vmin, vmax=vmax)
    rgba = plt.get_cmap(cmap)(norm(data))  # H×W×4 float [0,1]

    # Binary layers: 0-pixels → transparent; continuous → NaN → transparent
    is_binary = np.isin(np.unique(data[~np.isnan(data)]), [0, 1]).all()
    rgba[..., 3] = np.where(data == 1, opacity, 0) if is_binary else np.where(np.isnan(data), 0, opacity)

    buf = io.BytesIO()
    Image.fromarray((rgba * 255).astype(np.uint8)).save(buf, format="PNG")
    b64 = base64.b64encode(buf.getvalue()).decode()

    return folium.raster_layers.ImageOverlay(
        image=f"data:image/png;base64,{b64}",
        bounds=[[float(da.latitude.min()), float(da.longitude.min())],
                [float(da.latitude.max()), float(da.longitude.max())]],
        name=name,
        opacity=1.0,  # alpha already baked into RGBA
        show=True,
    )
```

```python
# ============================================================================
# PART A — Quantitative comparison of VH vs VV flood masks
# ============================================================================
# Your notebook computed `difference_binary.vh` (VH flood) and
# `difference_binary_vv.vv` (VV flood). Let's compute area and agreement.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Extract the 2D numpy arrays (binary: 1 = flood)
vh_mask = difference_binary.vh.values.astype(np.uint8)
vv_mask = difference_binary_vv.vv.values.astype(np.uint8)

# --- Per-pixel area in km² ---
# At EPSG:4326, a pixel's physical area depends on latitude. We compute the
# true area by taking the cosine-latitude-adjusted degrees-to-meters conversion.
lat_center = (latmin + latmax) / 2
pixel_deg = dx  # 0.0003°
# 1° latitude ≈ 111.32 km; 1° longitude ≈ 111.32 * cos(lat) km
pixel_area_km2 = (pixel_deg * 111.32) * (pixel_deg * 111.32 * np.cos(np.radians(lat_center)))
print(f'Each pixel ≈ {pixel_area_km2 * 1e6:.1f} m²  ({pixel_area_km2:.6f} km²)')

def flood_area_km2(mask):
    """Total flooded area in km² for a binary mask."""
    return int(mask.sum()) * pixel_area_km2

# --- Compute intersection/union between VH and VV ---
# Make sure they share the same grid — they should, both came from `dc`.
assert vh_mask.shape == vv_mask.shape, 'VH and VV shapes differ — fix upstream first'

both   = (vh_mask == 1) & (vv_mask == 1)
either = (vh_mask == 1) | (vv_mask == 1)
only_vh = (vh_mask == 1) & (vv_mask == 0)
only_vv = (vv_mask == 1) & (vh_mask == 0)

print('\n=== SAR flood extent comparison ===')
print(f'  VH only threshold (-1.5 dB):  {flood_area_km2(vh_mask):8.2f} km²  ({int(vh_mask.sum()):>8,} px)')
print(f'  VV only threshold (-1.5 dB):  {flood_area_km2(vv_mask):8.2f} km²  ({int(vv_mask.sum()):>8,} px)')
print(f'  VH ∩ VV (agree flooded):      {flood_area_km2(both):8.2f} km²  ({int(both.sum()):>8,} px)')
print(f'  VH ∪ VV (either flagged):     {flood_area_km2(either):8.2f} km²  ({int(either.sum()):>8,} px)')
print(f"  VH flagged, VV didn't:        {flood_area_km2(only_vh):8.2f} km²  ({int(only_vh.sum()):>8,} px)")
print(f"  VV flagged, VH didn't:        {flood_area_km2(only_vv):8.2f} km²  ({int(only_vv.sum()):>8,} px)")
iou_vh_vv = int(both.sum()) / max(int(either.sum()), 1)
print(f'  IoU (VH ∩ VV) / (VH ∪ VV):    {iou_vh_vv:.3f}')
```

```python
# ============================================================================
# MNDWI is computed on a slightly different Sentinel-2 grid (same bbox and dx,
# but from a separate STAC query). Align it to the SAR grid for pixel-level
# comparison.
# ============================================================================

# Rebuild the MNDWI flood mask cleanly (your notebook defined it as `flood` in cell 31).
# This is water_after AND NOT water_before.
mndwi_flood_mask = flood.astype(np.uint8)  # xarray DataArray with lat/lon coords

# Re-grid to match SAR mask. Both are EPSG:4326 at dx=0.0003 with the same bbox,
# so in principle they should already align — but odc-stac sometimes returns
# arrays that differ by one pixel at the edges. Force alignment:
import rioxarray
# Set CRS on both if not already set.
for da in (difference_binary.vh, mndwi_flood_mask):
    if not da.rio.crs:
        da.rio.write_crs('EPSG:4326', inplace=True)

# Reproject MNDWI flood onto the SAR flood grid. reproject_match handles
# any sub-pixel misalignment via nearest-neighbor.
mndwi_flood_aligned = mndwi_flood_mask.rio.reproject_match(difference_binary.vh)
mndwi_mask = mndwi_flood_aligned.values.astype(np.uint8)

print(f'\nMNDWI flood:  {flood_area_km2(mndwi_mask):8.2f} km²  ({int(mndwi_mask.sum()):>8,} px)')

# Method-by-method IoU table
def iou(a, b):
    inter = int(((a == 1) & (b == 1)).sum())
    union = int(((a == 1) | (b == 1)).sum())
    return inter / max(union, 1)

methods = {'VH': vh_mask, 'VV': vv_mask, 'MNDWI': mndwi_mask}
print('\n=== IoU matrix (binary flood masks) ===')
iou_rows = []
for name_a, a in methods.items():
    row = {'method': name_a}
    for name_b, b in methods.items():
        row[name_b] = iou(a, b)
    iou_rows.append(row)
print(pd.DataFrame(iou_rows).set_index('method').round(3).to_string())
```

```python
# Side-by-side flood masks + agreement map.
extent = [lonmin, lonmax, latmin, latmax]

fig, axes = plt.subplots(2, 2, figsize=(14, 13))

# Three individual masks
axes[0, 0].imshow(vh_mask,    origin='upper', extent=extent, cmap='Blues',   interpolation='none')
axes[0, 0].set_title(f'SAR VH  ({flood_area_km2(vh_mask):.1f} km²)')

axes[0, 1].imshow(vv_mask,    origin='upper', extent=extent, cmap='Reds',    interpolation='none')
axes[0, 1].set_title(f'SAR VV  ({flood_area_km2(vv_mask):.1f} km²)')

axes[1, 0].imshow(mndwi_mask, origin='upper', extent=extent, cmap='Greens',  interpolation='none')
axes[1, 0].set_title(f'MNDWI   ({flood_area_km2(mndwi_mask):.1f} km²)')

# Agreement: count how many methods flag each pixel (0-3)
n_agree = vh_mask + vv_mask + mndwi_mask
im = axes[1, 1].imshow(n_agree, origin='upper', extent=extent,
                       cmap='viridis', vmin=0, vmax=3, interpolation='none')
axes[1, 1].set_title('Method agreement (0=none, 3=all three)')
plt.colorbar(im, ax=axes[1, 1], shrink=0.7)

for ax in axes.flat:
    ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')

plt.suptitle('Baseline methods — VH vs VV vs MNDWI', y=1.00, fontsize=14)
plt.tight_layout()
plt.show()
```

```python

# ============================================================================
# PART B — Load ML (RF) flood TIFFs and compare to baselines
# ============================================================================
import rioxarray

# Adjust this path if needed
ML_TIFF_DIR = '/content/drive/MyDrive/Musa_650/Final/nairobi_flood_outputs'

def load_flood_tiff(path, template_da):
    """Load a flood GeoTIFF and reproject onto the baseline grid."""
    arr = rioxarray.open_rasterio(path).squeeze()
    # The TIFFs are uint8 in EPSG:32737 at 10m. Reproject to match the SAR grid.
    arr_aligned = arr.rio.reproject_match(template_da)
    return arr_aligned.astype(np.uint8)

template = difference_binary.vh  # baseline grid (EPSG:4326, 30m)

rf_conservative = load_flood_tiff(f'{ML_TIFF_DIR}/flood_conservative.tif',    template)
rf_sensitive    = load_flood_tiff(f'{ML_TIFF_DIR}/flood_sensitive.tif',       template)
rf_agree        = load_flood_tiff(f'{ML_TIFF_DIR}/flood_high_confidence.tif', template)

# Add to methods dict for full comparison
methods['RF conservative'] = rf_conservative.values
methods['RF sensitive']    = rf_sensitive.values
methods['RF agree']        = rf_agree.values

# Summary table of all methods
print('\n=== All methods — flood area summary ===')
summary = []
for name, mask in methods.items():
    n_px = int(mask.sum())
    summary.append({
        'method': name,
        'pixels': n_px,
        'area_km2': n_px * pixel_area_km2,
    })
summary_df = pd.DataFrame(summary).set_index('method').round({'area_km2': 2})
print(summary_df.to_string())
```

```python
import rioxarray
import numpy as np

ML_TIFF_DIR = '/content/drive/MyDrive/Musa_650/Final/nairobi_flood_outputs'

# Use the 10m RF grid as the common comparison grid
template_10m = rioxarray.open_rasterio(f'{ML_TIFF_DIR}/flood_conservative.tif').squeeze()

def to_10m_grid(da_30m, template=template_10m):
    if not da_30m.rio.crs:
        da_30m.rio.write_crs('EPSG:4326', inplace=True)
    return da_30m.rio.reproject_match(template).astype(np.uint8)

# Upsample baseline 30m masks to 10m (nearest-neighbor, each 30m pixel → 3×3 block)
vh_10m    = to_10m_grid(difference_binary.vh)
vv_10m    = to_10m_grid(difference_binary_vv.vv)
mndwi_10m = to_10m_grid(flood.astype(np.uint8))

# Load RF masks (already at 10m)
rf_c_10m = rioxarray.open_rasterio(f'{ML_TIFF_DIR}/flood_conservative.tif').squeeze().values
rf_s_10m = rioxarray.open_rasterio(f'{ML_TIFF_DIR}/flood_sensitive.tif').squeeze().values

methods_10m = {
    'VH':              vh_10m.values,
    'VV':              vv_10m.values,
    'MNDWI':           mndwi_10m.values,
    'RF conservative': rf_c_10m,
    'RF sensitive':    rf_s_10m,
}

print('methods_10m loaded:')
for name, mask in methods_10m.items():
    print(f'  {name:<18}  {int(mask.sum()):>10,} px  ({mask.sum() * 1e-4:.2f} km²)')
```

```python
# ============================================================================
# 2. Rasterize UNOSAT polygons onto the comparison grid
# ============================================================================
# Use the 10m RF grid since it preserves the most detail.
template = rioxarray.open_rasterio(
    f'{ML_TIFF_DIR}/flood_conservative.tif'
).squeeze()

# Rasterize UNOSAT polygons in the template's CRS (UTM 37S, 10m).
unosat_utm = unosat_clipped.to_crs(template.rio.crs)
unosat_mask = rasterize(
    [(geom, 1) for geom in unosat_utm.geometry if geom is not None and not geom.is_empty],
    out_shape=template.shape,
    transform=template.rio.transform(),
    fill=0,
    dtype=np.uint8,
)

print(f'UNOSAT raster: {unosat_mask.sum():,} pixels = {unosat_mask.sum() * 1e-4:.2f} km²')

# ============================================================================
# 3. Score every method against UNOSAT
# ============================================================================
# Reuse methods_10m from earlier (all at 10m on the RF grid).
# If you don't have it yet, rebuild by upsampling the baselines to 10m first.

pixel_area_km2_10m = 0.0001  # 10m × 10m = 100 m² = 0.0001 km²

def score_vs_reference(pred, ref):
    p = pred.flatten(); r = ref.flatten()
    return {
        'area_km2':  int(pred.sum()) * pixel_area_km2_10m,
        'precision': precision_score(r, p, zero_division=0),
        'recall':    recall_score(r,  p, zero_division=0),
        'f1':        f1_score(r,      p, zero_division=0),
        'iou':       jaccard_score(r, p, zero_division=0),
    }

rows = []
for name, mask in methods_10m.items():
    r = score_vs_reference(mask, unosat_mask)
    r['method'] = name
    rows.append(r)

# Also include UNOSAT's own area as reference.
unosat_row = {
    'method':    'UNOSAT (reference)',
    'area_km2':  unosat_mask.sum() * pixel_area_km2_10m,
    'precision': 1.0, 'recall': 1.0, 'f1': 1.0, 'iou': 1.0,
}
rows.append(unosat_row)

scores_df = pd.DataFrame(rows).set_index('method')[['area_km2', 'precision', 'recall', 'f1', 'iou']]
print('\n=== Scoring vs. UNOSAT reference ===')
print(scores_df.round(3).to_string())
```

```python
# ============================================================================
# 5. Spatial disagreement maps — where each method and UNOSAT diverge
# ============================================================================
from matplotlib.colors import ListedColormap
from matplotlib.patches import Patch

n = len(methods_10m)
fig, axes = plt.subplots(n, 1, figsize=(14, 10 * n))
if n == 1:
    axes = [axes]

# 0 = neither, 1 = method only (false positive), 2 = UNOSAT only (missed), 3 = both agree
cmap = ListedColormap(['#111111', '#FF3333', '#FFFF00', '#00FF00'])

for ax, (name, mask) in zip(axes, methods_10m.items()):
    agreement = np.zeros_like(mask, dtype=np.uint8)
    agreement[(mask == 1) & (unosat_mask == 0)] = 1
    agreement[(mask == 0) & (unosat_mask == 1)] = 2
    agreement[(mask == 1) & (unosat_mask == 1)] = 3
    ax.imshow(agreement, cmap=cmap, vmin=0, vmax=3, interpolation='nearest')
    ax.set_title(f'{name}   |   IoU = {scores_df.loc[name, "iou"]:.3f}   |   '
                 f'Area: {scores_df.loc[name, "area_km2"]:.1f} km²   vs.   UNOSAT: 55.2 km²',
                 fontsize=14)
    ax.axis('off')

legend = [
    Patch(facecolor='#00FF00', label='Both agree (flood)'),
    Patch(facecolor='#FF3333', label='Method only (false positive vs UNOSAT)'),
    Patch(facecolor='#FFFF00', label='UNOSAT only (missed)'),
    Patch(facecolor='#111111', label='Both agree (no flood)'),
]
fig.legend(handles=legend, loc='lower center', ncol=4, frameon=False,
           bbox_to_anchor=(0.5, 0.005), fontsize=12)
plt.suptitle('Spatial agreement: each method vs. UNOSAT flood extent', y=0.995, fontsize=16)
plt.tight_layout()
plt.subplots_adjust(bottom=0.03)
plt.show()
```
